Notebook này tái tạo các hình kết quả thực nghiệm RQ1–RQ3 từ **đúng các số liệu được in trong báo cáo nghiên cứu**. Không chạy lại benchmark và không tự tạo dữ liệu mới.

**Nguồn:** `Bao_cao_nghien_cuu_LLM_CP_SAT.docx`, phần 4 – Kết quả và Thảo luận, đặc biệt Bảng 1–3 và các hình ở trang 12–15.

**Lưu ý:** số liệu được nhập thủ công từ bảng trong báo cáo. Nếu sửa báo cáo, hãy cập nhật các bảng dữ liệu trong notebook rồi chạy lại. Các biểu đồ là bản tái tạo gần với bản gốc (Matplotlib), không phải ảnh chụp hoặc bản sao pixel-perfect.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display, Markdown

# Thiết lập hiển thị gần với biểu đồ trong báo cáo gốc
plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 200,
    "font.size": 9,
    "axes.titlesize": 10,
    "axes.labelsize": 9,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
})
print('Đã nạp pandas, matplotlib và numpy.')
```

Ba bảng dưới đây được tái nhập theo nội dung báo cáo. Dấu `—` được giữ nguyên khi báo cáo không cung cấp giá trị so sánh tương ứng.

```python
# Bảng 1 — Kết quả RQ1 của kiến trúc Hybrid (n = 60)
rq1_table = pd.DataFrame([
    ('End-to-end success', '100%'),
    ('Exact match – tags', '90.00%'),
    ('Exact match – constraints', '98.33%'),
    ('Exact match – RequirementSet', '88.33%'),
    ('Mean tag F1', '90.00%'),
    ('Constraint precision / recall / F1', '99.17% / 99.17% / 99.17%'),
    ('Feasibility detection accuracy', '100%'),
    ('Candidate-count exact', '90.00%'),
    ('Multi-turn context action accuracy', '100% (8/8)'),
], columns=['Chỉ số', 'Kết quả'])
display(rq1_table)
```

## 2. Hình RQ1 — Độ chính xác của tầng diễn giải yêu cầu

Biểu đồ dùng 7 chỉ số được thể hiện trong hình RQ1 của báo cáo. Các giá trị phần trăm được giữ theo số liệu trong Bảng 1. Đây là các metric khác nhau, không nên hiểu chúng là một phép đo duy nhất.

```python
rq1_plot = pd.DataFrame({
    'Chỉ số': [ 'Constraint exact', 'Constraint F1', 'Feasibility accuracy'],
    'Tỷ lệ (%)': [ 88.33 , 99.17, 100.0],
})
fig, ax = plt.subplots(figsize=(9, 4.8))
bars = ax.bar(rq1_plot['Chỉ số'], rq1_plot['Tỷ lệ (%)'])
ax.set_title('RQ1 – Độ chính xác của tầng diễn giải yêu cầu (Hybrid, n=60)')
ax.set_ylabel('Tỷ lệ (%)')
ax.set_ylim(0, 110)
ax.grid(axis='y', alpha=0.25)
ax.set_axisbelow(True)
ax.tick_params(axis='x', rotation=28)
for bar, val in zip(bars, rq1_plot['Tỷ lệ (%)']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 1.2, f'{val:.1f}%', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()

display(Markdown('**Cách đọc:** các cột biểu diễn tỷ lệ chính xác theo từng tiêu chí. Constraint exact/F1 cao hơn tag exact và RequirementSet exact; điều này phù hợp với nhận xét trong báo cáo rằng suy luận nhãn ngữ nghĩa còn khó hơn trích xuất giá trị định lượng.'))
```

## 3. Hình RQ2 — Tính hợp lệ của sản phẩm trong danh mục

Hình này so sánh catalog validity của ba hệ thống trên 52 comparison cases.

```python
catalog_validity = pd.DataFrame({
    'Hệ thống': ['LLM-only', 'Catalog-grounded LLM', 'Hybrid'],
    'Catalog validity (%)': [0, 100, 100],
})
fig, ax = plt.subplots(figsize=(7.5, 4.6))
bars = ax.bar(catalog_validity['Hệ thống'], catalog_validity['Catalog validity (%)'])
ax.set_title('RQ2 – Tính hợp lệ của sản phẩm trong danh mục')
ax.set_ylabel('Tỉ lệ sản phẩm sau dự đoán nằm trong dữ liệu yêu cầu của các hệ thốn (%)')
ax.set_ylim(0, 110)
ax.grid(axis='y', alpha=0.25)
ax.set_axisbelow(True)
for bar, val in zip(bars, catalog_validity['Catalog validity (%)']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 1.5, f'{val:.0f}%', ha='center', va='bottom')
plt.tight_layout()
plt.show()

display(Markdown('**Cách đọc:** LLM-only có 0% catalog validity, trong khi hai hệ thống sử dụng catalog grounding đều đạt 100%. Không được diễn giải 0% catalog validity thành 100% hallucination; báo cáo phân biệt sản phẩm ngoài catalog với sản phẩm không tồn tại.'))
```

## 4. Hình RQ2 — Độ tin cậy ràng buộc trên cùng catalog

Biểu đồ gốc chỉ so sánh Catalog-grounded LLM và Hybrid ở hai metric: all-hard satisfaction và soft violation. Không bổ sung LLM-only vào hình này vì bảng báo cáo không cung cấp all-hard satisfaction so sánh tương đương cho hệ đó.

```python
constraint_reliability = pd.DataFrame({
    'Hệ thống': ['Catalog-grounded LLM', 'Hybrid'],
    'All-hard satisfaction (%)': [72.73, 100.0],
    'Soft violation (%)': [13.33, 6.67],
})
x = np.arange(len(constraint_reliability))
width = 0.34
fig, ax = plt.subplots(figsize=(7.8, 4.8))
bars1 = ax.bar(x - width/2, constraint_reliability['All-hard satisfaction (%)'], width, label='mức độ thoả mãn ràng buộc cứng (%)')
bars2 = ax.bar(x + width/2, constraint_reliability['Soft violation (%)'], width, label='mức độ vi phạm ràng buộc mềm  (%)')
ax.set_title('RQ2 – Độ tin cậy ràng buộc trên cùng catalog')
ax.set_ylabel('Tỷ lệ (%)')
ax.set_xticks(x, constraint_reliability['Hệ thống'])
ax.set_ylim(0, 110)
ax.legend(loc='upper left', frameon=True)
ax.grid(axis='y', alpha=0.25)
ax.set_axisbelow(True)
for bars in [bars1, bars2]:
    for bar in bars:
        val = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, val + 1.2, f'{val:.2f}%', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()

display(Markdown('**Cách đọc:** Hybrid đạt 100% all-hard satisfaction trong các trường hợp được đánh giá và có soft violation 6,67%, thấp hơn mức 13,33% của Catalog-grounded LLM. So sánh này không cô lập riêng tác động nhân quả của CP-SAT vì kiến trúc Hybrid còn có retrieval, RequirementSet và scoring.'))
```

## 5. Hình RQ3 — Độ trễ trung vị

Vẽ median latency theo Bảng 3. Báo cáo chọn trung vị để tránh việc một outlier khoảng 354,7 giây làm mean của LLM-only tăng mạnh.

```python
latency_median = pd.DataFrame({
    'Hệ thống': ['LLM-only', 'Catalog-grounded LLM', 'Hybrid'],
    'Median latency (s)': [1.635, 2.097, 1.617],
})
fig, ax = plt.subplots(figsize=(7.5, 4.6))
bars = ax.bar(latency_median['Hệ thống'], latency_median['Median latency (s)'])
ax.set_title('RQ3 – Độ trễ trung vị')
ax.set_ylabel('Giây')
ax.set_ylim(0, 2.4)
ax.grid(axis='y', alpha=0.25)
ax.set_axisbelow(True)
ax.tick_params(axis='x', rotation=8)
for bar, val in zip(bars, latency_median['Median latency (s)']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.04, f'{val:.3f}s', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()

display(Markdown('**Cách đọc:** median của LLM-only và Hybrid gần tương đương (1,635 giây và 1,617 giây). Vì vậy, số liệu này không hỗ trợ kết luận Hybrid nhanh hơn nhiều lần.'))
```

## 6. Hình RQ3 — Token trung bình và chi phí API trung bình

Hai biểu đồ đặt cạnh nhau như trong báo cáo: tổng token trung bình mỗi truy vấn và chi phí API trung bình mỗi truy vấn.

```python
systems = ['LLM-only', 'Catalog-grounded LLM', 'Hybrid']
mean_total_tokens = [456.38, 9332.08, 1357.35]
mean_cost_usd = [0.000658, 0.003232, 0.000763]

fig, axes = plt.subplots(1, 2, figsize=(10, 4.3))
# Biểu đồ tổng token trung bình
bars = axes[0].bar(systems, mean_total_tokens)
axes[0].set_title('RQ3 – Mức sử dụng token trung bình')
axes[0].set_ylabel('Token trung bình / truy vấn')
axes[0].grid(axis='y', alpha=0.25)
axes[0].set_axisbelow(True)
axes[0].tick_params(axis='x', rotation=18)
for bar, val in zip(bars, mean_total_tokens):
    axes[0].text(bar.get_x() + bar.get_width()/2, val + max(mean_total_tokens)*0.025, f'{val:,.0f}', ha='center', va='bottom', fontsize=8)

# Biểu đồ chi phí API trung bình
bars = axes[1].bar(systems, mean_cost_usd)
axes[1].set_title('RQ3 – Chi phí API trung bình trên mỗi truy vấn')
axes[1].set_ylabel('USD / truy vấn')
axes[1].grid(axis='y', alpha=0.25)
axes[1].set_axisbelow(True)
axes[1].tick_params(axis='x', rotation=18)
for bar, val in zip(bars, mean_cost_usd):
    axes[1].text(bar.get_x() + bar.get_width()/2, val + max(mean_cost_usd)*0.025, f'{val:.6f}', ha='center', va='bottom', fontsize=7)
plt.tight_layout()
plt.show()

display(Markdown('**Cách đọc:** Catalog-grounded LLM dùng trung bình 9.332 token/truy vấn, cao hơn rõ rệt Hybrid (1.357 token). Chi phí trung bình lần lượt là 0,003232 USD và 0,000763 USD; theo báo cáo, Hybrid giảm khoảng 76,4% chi phí so với full-catalog prompting, nhưng cao hơn LLM-only khoảng 15,9%.'))
```
