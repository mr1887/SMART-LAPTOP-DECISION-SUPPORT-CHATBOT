```python
import pandas as pd
df = pd.read_csv("D:\Python project\dự án chatbot\data\processed\laptop_dataset.csv")
eng = pd.read_csv("D:\Python project\dự án chatbot\data\processed\laptop_engagement_features.csv")
df.head()
```

```python
# Tính số lượng và tỷ lệ % thiếu
missing_df = pd.DataFrame({
    'Số giá trị thiếu': df.isnull().sum(),
    'Tỷ lệ % thiếu': (df.isnull().sum() / len(df)) * 100
})

# Lọc ra những cột CÓ giá trị thiếu và sắp xếp giảm dần
missing_df = missing_df[missing_df['Số giá trị thiếu'] > 0].sort_values(by='Tỷ lệ % thiếu', ascending=False)

# Hiển thị bảng
print(missing_df)
```
